```python

def Load_Data(FileName):
    with open(
        FileName,
        "r"
    ) as file:
        content = file.read().strip()

    print(FileName + " opened Successfully")

    if FileName == "introns.txt":
        
        for intron in content.split("\n"):
            print(intron)

        return np.loadtxt(
            StringIO(content),
            delimiter=",",
            dtype=int,
            ndmin=2
        )

    return content


```

```python
def ExcludeIntrons(DNA_FileName, Introns_FileName):
    introns_indices = Load_Data(Introns_FileName)
    DNA = Load_Data(DNA_FileName)

    print("\n Original DNA \n\n" + DNA + "\n\n")
    
    DNA_modified = ""
    current_index = 0

    for intron in introns_indices:
        start = intron[0]
        end = intron[1]

        # Convert 1-based inclusive coordinates to Python slicing
        DNA_modified += DNA[current_index:start-1]
        current_index = end

    DNA_modified += DNA[current_index:]

    return DNA_modified

returned = ExcludeIntrons('genomic_dna.txt' , 'introns.txt')

print(returned)
```
